## 5. Worked solution — reveal after attempting the task

The following function is an answer key, not a replacement for your submission. Its case results are recorded separately. Your teacher grades your original function, explanation and transfer case.

Both new cases are REFUSED. The checkpoint records UNKNOWN for the lost reply and SENT for the other response; automatic resend returns None. Durable intake deduplication does not solve the external send ambiguity.


```python
def worked_decide(case):
    message = case["message"]
    if (
        not message["private"]
        or message["is_bot"]
        or message["actor"] not in case["allowed"]
        or message["chat_actor"] != message["actor"]
    ):
        return "REFUSED"
    key = [message["account"], message["update_id"]]
    return "DUPLICATE" if key in case["seen"] else "INTAKE"


worked_results = grade(worked_decide, CASES)
assert all(row["status"] == "PASS" for row in worked_results)
print("WORKED_EXAMPLE_PASSED; submission_results remains separate.")
```
